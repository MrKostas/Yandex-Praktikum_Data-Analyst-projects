### Преобразований данных в нужные типы

После вывода информации о датасете методом .info() в предыдущем блоке заметили, что столбец с годом релиза игры ('year_of_release'), а также столбцы 'na_sales', 'eu_sales', 'jp_sales', 'other_sales', которые хранят информацию о количестве проданных копий (в миллионах штук), содержат значения вещественного типа данных.

Так же отметим то, что числовая оценка пользователей ('user_score') является строковым типом данных.

Работе над этим столбцами будет посвящен этот блок.

```python
# заменим значения в столбце year_of_release на целочисленный тип
try:
    df['year_of_release'] = df['year_of_release'].astype('int')
except:
    print('''Невозможно заменить значения в столбце на целочисленный тип, когда есть пропуски''')
    print()
    print('Количество пропусков:', df['year_of_release'].isna().sum())
    print('Процент пропусков в столбце "year_of_release": {:.1%}'.format(df['year_of_release'].isna().sum() / df['year_of_release'].count()))
```

Так как процент пропусков в стобце ничтожно мал, можно избавиться от этих записей.
Вместе с этим, посмотрим пропуски по остальным столбцам датафрейма:

```python
# вызовем метод .isna() в комбинации с .sum(), чтобы посмотреть пропуски по столбцам
df.isna().sum()
```

```python
# удалим строки с пропускам, указав в аргументе subset названия столбцов с минимальным количеством пропусков
df = df.dropna(subset = ['name', 'year_of_release', 'genre'])
df.isna().sum()
```

Изменим тип данных в столбце 'user_score', так как сейчас он имеет тип 'object'

```python
# посмотрим, из-за чего столбец имеет строковый тип
print(df['user_score'].unique())
# Аббревиатура tbd значит to be determined, to be done
# То есть, данные были нарочно не заполнены, так как не определились с рейтингом. Заменим 'tbd' на NaN
df.loc[df['user_score'] == 'tbd', 'user_score'] = 'nan'
df['user_score'] = df['user_score'].astype('float')
```

### Замена пропусков

Так как пропуски имеют столбцы 'critic_score' ("оценка критиков"), 'user_score' ("оценка игроков") и 'rating' (рейтинг игры), и мы не имеем возможности самостоятельно присвоить играм значения оценок (нельзя поставить среднее или медианное значение по столбцу для каждой уникальной игры, так как это сильно исказит результаты анализа.

Оставим пропуски как есть, но в столбце 'rating' заменим их на значения 'unknown', что позволит выделить их в отдельную категорию.

```python
df['rating'] = df['rating'].fillna('unknown')
```

### Изучение корреляции оценок пользователей и критиков и общих продаж

Построим диаграммы рассеяния для ТОП-5 популярных платформ, которые отфильтровали за период с 2013 года по настоящее (на момент исследования) время:

```python
# напишем функцию для построения графиков и отображения корреляции для каждой платформы, чтобы не писать все вручную несколько раз
def platform_statistics(name_of_platform):
    platform = df_new[df_new['platform'] == name_of_platform]
    fig, ax = plt.subplots(1 ,2, figsize=(15,5))
    sns.scatterplot(x='user_score', y='total_sales', data=platform, ax=ax[0])
    sns.scatterplot(x='critic_score', y='total_sales', data=platform, ax=ax[1])
    fig.suptitle(name_of_platform, fontsize=15)
    ax[0].set(xlabel='Оценка пользователей')
    ax[1].set(xlabel='Оценка критиков')
    ax[0].set(ylabel='Суммарные продажи')
    ax[1].set(ylabel='Суммарные продажи')
    plt.show()
    
    user_correl = platform['user_score'].corr(platform['total_sales'])
    critic_correl = platform['critic_score'].corr(platform['total_sales'])
    print('Корреляция оценок игроков и продаж видеоигр:', user_correl)
    print('Корреляция оценки критиков и продаж видеоигр:', critic_correl)
```

## Проверка гипотез

### Средние пользовательские рейтинги платформ Xbox One и PC одинаковые

**Сформулируем гипотезы:**

* Нулевая гипотеза H₀: Средние пользовательские рейтинги платформ Xbox One и PC одинаковые;
* Альтернативная гипотеза H₁: Средние пользовательские рейтинги платформ Xbox One и PC различаются

<div class="alert alert-success"; style="border-left: 7px solid green">
<font size='4'><b>👍 Комментарий ревьюера</b></font>

Корректно сформулировал
</div>

```python
# Сохраним в переменных xbox_hyp и pc_hyp данные с 2005 года по настоящее время
xone_scores = df[(df['platform']=='XOne') & (df['year_of_release']>=2005)]['user_score']
pc_scores = df[(df['platform']=='PC') & (df['year_of_release']>=2005)]['user_score']

#Посчитаем средний рейтинг пользователя для xbox платформ
print(xone_scores.mean())
print(pc_scores.mean())
```

### Средние пользовательские рейтинги жанров Action и Sports разные

**Сформулируем гипотезы:**

* Нулевая гипотеза H₀: Средние пользовательские рейтинги жанров Action и Sports разные;
* Альтернативная гипотеза H₁: Средние пользовательские рейтинги жанров Action и Sports одинаковые;

```python
# Сохраним в переменных xbox_hyp и pc_hyp данные с 2005 года по настоящее время
sports_scores = df[(df['genre']=='Sports') & (df['year_of_release']>=2005)]['user_score']
action_scores = df[(df['genre']=='Action') & (df['year_of_release']>=2005)]['user_score']

#Посчитаем средний рейтинг пользователя для xbox платформ
print(sports_scores.mean())
print(action_scores.mean())
```

```python
#Выполним проверку гипотезы. Будем использовать метод ttest_ind

alpha = .01

results = st.ttest_ind(action_scores.dropna(), sports_scores.dropna(), equal_var=False)

print('p-значение:', results.pvalue)


if (results.pvalue < alpha):
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу")
```
