## Описание проекта

Сеть фитнес-центров «Культурист-датасаентист» разрабатывает стратегию взаимодействия с клиентами на основе аналитических данных.

Распространённая проблема фитнес-клубов и других сервисов — отток клиентов. Как понять, что клиент больше не с вами? Можно записать в отток тех, кто попросил закрыть договор или удалил аккаунт. Однако клиенты не всегда уходят демонстративно: чаще перестают пользоваться сервисом тихо.

Индикаторы оттока зависят от специфики отрасли. Когда пользователь редко, но стабильно закупается в интернет-магазине — не похоже, что он «отвалился». А вот если две недели не заходит на канал с ежедневно обновляемым контентом, дела плохи: подписчик заскучал и, кажется, оставил вас.

Для фитнес-центра можно считать, что клиент попал в отток, если за последний месяц ни разу не посетил спортзал. Конечно, не исключено, что он уехал на Бали и по приезде обязательно продолжит ходить на фитнес. Однако чаще бывает наоборот. Если клиент начал новую жизнь с понедельника, немного походил в спортзал, а потом пропал — скорее всего, он не вернётся.

Чтобы бороться с оттоком, отдел по работе с клиентами «Культуриста-датасаентиста» перевёл в электронный вид множество клиентских анкет. Наша задача — провести анализ и подготовить план действий по удержанию клиентов.

**Вопросы, которые необходимо решить в ходе проекта:**
* научиться прогнозировать вероятность оттока (на уровне следующего месяца) для каждого клиента;
* сформировать типичные портреты клиентов: выделить несколько наиболее ярких групп и охарактеризовать их основные свойства;
* проанализировать основные признаки, наиболее сильно влияющие на отток;
* сформулировать основные выводы и разработать рекомендации по повышению качества работы с клиентами:

    1. выделить целевые группы клиентов;
    2. предложить меры по снижению оттока;
    3. определить другие особенности взаимодействия с клиентами.

### Изучение средних значений и стандартных отклонений

```python
# вызовем метод describe() для изучения статистических данных датасета и транспонируем получившуюся
# таблицу
data.describe().T.sort_values('mean', ascending=False)
```

**Вывод**:
При транспонировании таблицы и сортировке поля со средним значением по убыванию становится видно, что все признаки делятся на бинарные (от 'phone' и ниже) и не бинарные (от 'avg_additional_charges_total' до 'phone').

Так как в описании датасета не содержится описания бинарных величин, то будем считать "0" и "1" как булевы значения, где "0" - это "нет", а "1" - это "да". В характеристике пола примем, что "1" - это мужчина, а "0" - это женщина. 

```python
# изучим средние значения по двум группам - те, кто ушел в отток, и те, кто остался - и транспонируем
# таблицу
churn_data = data.groupby('churn').agg('mean').T
# заменим названия полей таблицы на "churn_false" и "churn_true"
churn_data.columns = ['churn_false', 'churn_true']
# добавим поле с расчетом дельты, чтобы в дальнейшем отсортировать таблицу и увидеть самые важные
# признаки
churn_data['delta'] = (churn_data['churn_false'] - churn_data['churn_true']).abs()
# отсортируем таблицу
churn_data.sort_values(by='delta', ascending=False)
```

**Вывод:**
Таким образом, у нас получился ТОП признаков пользователей с ранжированием по дельте между ушедшими и оставшими клиентами. Выведем характеристику 5 признаков с самой большой дельтой:
1. **'avg_additional_charges_total'** - суммарная выручка от других услуг фитнес-центра: кафе, спорттовары, косметический и массажный салон;
2. **'contract_period'** - длительность текущего действующего абонемента (месяц, 6 месяцев, год);
3. **'lifetime'** - время с момента первого обращения в фитнес-центр (в месяцах);
4. **'month_to_end_contract'** - срок до окончания текущего действующего абонемента (в месяцах);
5. **'age'** - возраст;

```python
# строим графики с помощью цикла
for i in churn_data.index:
    plt.figure(figsize=(10, 5))
    sns.distplot(data[data['churn'] == 0][i], bins=10, label='churn_false')
    sns.distplot(data[data['churn'] == 1][i], bins=10, label='churn_true')
    plt.legend()
    plt.title(i, fontsize=20)
    plt.show()
```

```python
# Применяю стандартизацию к матрице признаков для тестовой выборки
X_test_st = scaler.transform(X_test)
```

```python
# воспользуемся функцией из теории для обучения моделей
def make_prediction(model, X_train,y_train,X_test,y_test):
    result = {}
    model = model
    model.fit(X_train, y_train)
    predict = model.predict(X_test)
    result[model] = {'Accuracy':round(accuracy_score(y_test, predict),3),
                    'Precision':round(precision_score(y_test, predict),3),
                    'Recall':round(recall_score(y_test, predict),3)}
    print(result)
```

### Кластеризация пользователей

```python
# будем использовать очищенный от полей 'month_to_end_contract' и 'avg_class_frequency_total' датасет
# заново введем переменную scaler
scaler = StandardScaler()
# выполним обучение нормализатора и удалим поле с целевой переменной
X_sc = scaler.fit_transform(data.drop('churn', axis=1))
```
