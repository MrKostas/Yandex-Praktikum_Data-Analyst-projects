# Определение перспективного тарифа для телеком-компании

## Описание проекта

По условию задачи мы являемся аналитиком компании «Мегалайн» — федерального оператора сотовой связи.

Клиентам предлагают два тарифных плана: «Смарт» и «Ультра». Чтобы скорректировать рекламный бюджет, коммерческий департамент хочет понять, какой тариф приносит больше денег.

Необходимо будет сделать предварительный анализ тарифов на небольшой выборке клиентов. В распоряжении данные 500 пользователей «Мегалайна»: кто они, откуда, каким тарифом пользуются, сколько звонков и сообщений каждый отправил за 2018 год. Нужно проанализировать поведение клиентов и сделать вывод — какой тариф лучше.

## Вводная информация

**Тариф «Смарт»**
1. Ежемесячная плата: 550 рублей
2. Включено 500 минут разговора, 50 сообщений и 15 Гб интернет-трафика
3. Стоимость услуг сверх тарифного пакета:
    * минута разговора: 3 рубля
    * сообщение: 3 рубля
    * 1 Гб интернет-трафика: 200 рублей
    
**Тариф «Ультра»**
1. Ежемесячная плата: 1950 рублей
2. Включено 3000 минут разговора, 1000 сообщений и 30 Гб интернет-трафика
3. Стоимость услуг сверх тарифного пакета:
    * минута разговора: 1 рубль
    * сообщение: 1 рубль
    * 1 Гб интернет-трафика: 150 рублей
    
**Примечание**  
«Мегалайн» всегда округляет секунды до минут, а мегабайты — до гигабайт. Каждый звонок округляется отдельно: даже если он длился всего 1 секунду, будет засчитан как 1 минута.
Для веб-трафика отдельные сессии не считаются. Вместо этого общая сумма за месяц округляется в бо́льшую сторону. Если абонент использует 1025 мегабайт в этом месяце, с него возьмут плату за 2 гигабайта.

В датасете с информацией об интернет сессиях пользователей присутствует столбец "Unnamed: 0". Данного столбца нет в описании задания, а его значения дублируют индексы строк. Следовательно, надобности в данном столбце нет - его можно удалить:

```python
# применим метод .drop()
internet_df = internet_df.drop('Unnamed: 0', 1)
display(internet_df)
```

Бросаются в глаза столбцы, содержащие значения даты. Переведем значения данного столбца в формат "даты": 

```python
# применим метод .to_datetime()
calls_df['call_date'] = pd.to_datetime(calls_df['call_date'], format='%Y-%m-%d')
internet_df['session_date'] = pd.to_datetime(internet_df['session_date'], format='%Y-%m-%d')
messages_df['message_date'] = pd.to_datetime(messages_df['message_date'], format='%Y-%m-%d')
users_df['reg_date'] = pd.to_datetime(users_df['reg_date'], format='%Y-%m-%d')
```

Так как операторы округляют использованные минуты и мегабайты в большую сторону, то округлим и мы до большего целого числа:

```python
# добавим столбец с месяцем в датафреймы 'calls_df' и 'internet_df'
calls_df['month'] = calls_df['call_date'].dt.month
internet_df['month'] = internet_df['session_date'].dt.month
# добавим столбец с использованными гигабайтами в датафрейм 'internet_df'
internet_df['gb_used'] = internet_df['mb_used'] / 1024
# вызовем метод np.ceil() для округления значений использованных мегабайт и минут
calls_df['duration'] = np.ceil(calls_df['duration']).astype('Int64')
internet_df['gb_used'] = np.ceil(internet_df['gb_used']).astype('Int64')
```

### Проведение расчетов

#### Расчет количества звонков и израсходованных минут по месяцам для каждого пользователя

Необходимо посчитать для каждого пользователя количество звонков по месяцам. Создадим сводную таблицу с отображением количества звонков в каждом месяце:

```python
# для отображения количества звонков каждого пользователя, создадим сводную таблицу с id пользователя и месяцами по строкам
# заменим пустые значения в сводной таблице (пользователь не совершал звонков в конкретные месяцы) на нули
calls_pivot = calls_df.pivot_table(index = ['user_id','month'], values = 'call_date', aggfunc='count').fillna(0)
calls_pivot = calls_pivot.rename(columns={'call_date' : 'calls_quantity'})
display(calls_pivot)
```

```python
# заменим пустые значения в сводной таблице 'calls_pv' (пользователь не совершал звонков в конкретные месяцы) на нули
calls_pv = calls_pv.fillna(0)
display(calls_pv)
```

```python
# совместим две таблицы в одну методом .merge()
calls_full_pv = calls_pv.merge(calls_pivot, on=['user_id', 'month'], how='outer')
display(calls_full_pv)
```

#### Расчет количества отправленных сообщений и объемов израсходованного интернет-траффика

По такому же принципу создадим остальные таблицы с количеством отправленных сообщений, израсходованным пакетом интернет-трафика по месяцам и совместим все в одну сводную таблицу:

```python
# делаем сводную таблицу по количеству отправленных сообщений по месяцам
messages_df['month'] = pd.DatetimeIndex(messages_df['message_date']).month
messages_pv = (messages_df
    .pivot_table(index = ['user_id','month'], values = 'id', aggfunc='count')
    .fillna(0)
    .rename(columns={'id':'sent_messages'})
)
```

После проделанных действий совместим все в один готовый датафрейм:

```python
def total_price(row):
    if row['tariff'] == 'smart':
        if row['sent_messages'] > 50:
            message_extra_price = (row['sent_messages'] - 50) * 3
        else:
            message_extra_price = 0
        if row['duration'] > 500:
            calls_extra_price = (row['duration'] - 500) * 3
        else:
            calls_extra_price = 0
        if row['gb_used'] > 15:
            internet_extra_price = (row['gb_used'] - 15) * 200
        else:
            internet_extra_price = 0
        total_account = message_extra_price + calls_extra_price + internet_extra_price + 550
        
    if row['tariff'] == 'ultra':
        if row['sent_messages'] > 1000:
            message_extra_price = (row['sent_messages'] - 1000) * 1
        else:
            message_extra_price = 0
        if row['duration'] > 3000:
            calls_extra_price = (row['duration'] - 3000) * 1
        else:
            calls_extra_price = 0
        if row['gb_used'] > 30:
            internet_extra_price = (row['gb_used'] - 30) * 150
        else:
            internet_extra_price = 0
        total_account = message_extra_price + calls_extra_price + internet_extra_price + 1950
    return total_account
```

Поясню, харкод, это когда ты подставляешь какие-то значения в код руками, а не вычисляешь методами из данных.
