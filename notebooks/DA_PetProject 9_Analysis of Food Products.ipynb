```python
# Воспользуемся Z-критерием
def z_test(exp1, exp2, event, alpha): 
    p1_ev = all_funnels.loc[event, exp1]
    p2_ev = all_funnels.loc[event, exp2] 
    p1_us = users.loc[exp1, 'device_id_hash'] 
    p2_us = users.loc[exp2, 'device_id_hash'] 
    p1 = p1_ev / p1_us 
    p2 = p2_ev / p2_us 
    difference = p1 - p2
    p_combined = (p1_ev + p2_ev) / (p1_us + p2_us) 
    z_value = difference / math.sqrt(p_combined * (1 - p_combined) * (1 / p1_us + 1 / p2_us))
    distr = st.norm(0, 1)
    p_value = (1 - distr.cdf(abs(z_value))) * 2
    print('Проверка для  {} и {}, событие: {}, p-значение: {p_value:.2f}'.format(exp1, exp2, event, p_value=p_value))
    if (p_value < alpha):
        print("Отвергаем нулевую гипотезу")
    else:
        print("Не получилось отвергнуть нулевую гипотезу")
```

**Сформулируем гипотезы**

* Н0 - различия между экспериментальными группами, совершившими указанное событие, статистически не значимы

* Н1 - различия между экспериментальными группами, совершившими указанное событие, статистически значимы

Пороговое значение alpha примем за 5%.

Здесь можно и даже нужно применить поправку на множественную проверку гипотез.<br>
    
Во время проверки статистических гипотез при отвержении основной гипотезы (H0) возможна ошибка (ложное отклонение гипотезы, ошибка первого рода). Вероятность такого события ограничивается неким малым предварительно выбранным значением — уровнем значимости обычно alpha = 0.05.<br>    
    
Если всякий раз вероятность ошибиться равна ɑ, то вероятность не ошибиться: 1-ɑ. Так, вероятность не ошибиться ни разу за N сравнений равна: (1-ɑ)^N <br>

В итоге, вероятность ошибиться хотя бы раз за N сравнений: 1 - (1-ɑ)^N  <br>
    
За N в нашем случае берем 12, т.к. мы можем учитывать 12 тестов (А1/А2, А1/В, А2/В), а 4 теста (А1+А2)/В не учитываем для поправки, потому как А1/В, А2/В уже рассмотрели.<br> 
    
Следовательно, если мы применим поправку Бонферрони, альфа будет равна 0.05/12 = 0.004.<br>
    
Вероятность ошибки в нашем случае **без поправки, т.е. с ɑ=0,05**: 1-(1-0.05)^12= 46% <br>
Итак, с вероятностью появления ошибки в 46% проводить тестирование не очень целесообразно :)  <br>  
    
Но мы можем скорректировать уровень значимости, для этого используем один из базовых методов - метод Бонферрони.<br>    

Вероятность ошибки в этом случае, т.е. **с учетом поправки, т.е. с ɑ=0,004**: 1-(1-0.004)^12= 5% <br>
    
Можно было повторить все тесты с такой альфой, но в данном случае в этом нет необходимости.<br>
Обрати внимание, в проведенных тестах без поправки Бонферрони p_value везде больше 0,05, в большинстве случае значительно.
Это нам подсказывает, что с поправкой Бонферони будет также)
