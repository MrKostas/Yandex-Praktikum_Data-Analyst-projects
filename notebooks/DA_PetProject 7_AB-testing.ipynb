## Приоритезация гипотез <a id="prioritization"></a>

Исходя из вводных условий, на данном этапе необходимо выполнить следующие шаги:
- Применить фреймворк ICE для приоритизации гипотез;
- Применить фреймворк RICE для приоритизации гипотез;
- Указать, как изменилась приоритизация гипотез при применении RICE вместо ICE.

Оба фреймворка необходимо отсортировать по убыванию их приоритета.

Фреймворк **ICE** расшифровывается как impact, confidence, effort / ease «влияние, уверенность, усилия / простота».

**RICE** -  reach, impact, confidence, effort / ease «влияние, уверенность, усилия, охват / простота».

- *Reach* — скольких пользователей затронет изменение, которое вы хотите внести;
- *Impact* — насколько сильно изменение повлияет на пользователей, их опыт и удовлетворение от продукта;
- *Confidence* — насколько вы уверены, что это изменение повлияет на пользователей именно так;
- *Efforts* — сколько стоит протестировать эту гипотезу.

```python
# создадим в датасете новое поле, где будет рассчитан ICE
data['ICE'] = round(data['confidence'] * data['impact'] / data['efforts'], 2)
display(data.sort_values(by='ICE', ascending = False))
```

```python
# добавим в датасет новое поле, где будет рассчитан RICE
data['RICE'] = round(data['confidence'] * data['impact'] * data['reach'] / data['efforts'], 2)
display(data.sort_values(by='RICE', ascending = False))
```

**Вывод**:

- Применили фреймворк 'ICE' для приоритезации гипотез;
- Применили фреймворк 'RICE' для приоритезации гипотез;

Положение гипотез меняется при изменении сортировки по столбцу 'ICE' на сортировку по столбцу 'RICE'. Напомним, что фреймворк 'RICE' учитывает охват пользователей, которое затронет вводимое изменение, в отличие от 'ICE'.

При изменении сортировки можно заметить, как в первой тройке гипотез в двух случаях находятся гипотезы:
- **Добавить форму подписки на все основные страницы, чтобы собрать базу клиентов для email-рассылок**;
- **Добавить два новых канала привлечения трафика, что позволит привлекать на 30% больше пользователей**;

Мощность этих гипотез высока в обоих фреймворках, и, следовательно, необходимо уделить им внимание в первую очередь.

А вот гипотеза **'Запустить акцию, дающую скидку на товар в день рождения'** имеет максимальную мощность при проверке 'ICE', но в фреймворке 'RICE' выбывает из первой тройки по причине малого охвата пользователей.

## Анализ A/B-теста <a id="analysis"></a>

```python
# прочтем и запишем в переменные датасеты с результатами тестирования
orders = pd.read_csv('/datasets/orders.csv')
visitors = pd.read_csv('/datasets/visitors.csv')
# выведем общую информацию о датасетах
print(orders.info())
print(visitors.info())
```

```python
# исправим тип данных поля 'date' в датасете 'orders' на тип datetime
orders['date'] = pd.to_datetime(orders['date'])
# проверим тип данных столбца датасета
orders.info()
```

### Построение графика кумулятивной выручки по группам

```python
# создадим переменную, где будут уникальные пары значений групп и дат
datesGroups = orders[['date', 'group']].drop_duplicates()
```

```python
# объявляем новую переменную, которая будет содержать дату; группу теста; суммарную выручку; число уникальных пользователей,
# совершивших хотя бы один заказ; число уникальных заказов

ordersAggregated = datesGroups.apply(
    lambda x: orders[
        np.logical_and(
            orders['date'] <= x['date'], orders['group'] == x['group']
        )
    ].agg(
        {
            'date': 'max',
            'group': 'max',
            'transactionId': 'nunique',
            'visitorId': 'nunique',
            'revenue': 'sum',
        }
    ),
    axis=1,
).sort_values(by=['date', 'group']) # отсортируем столбцы 'date' и 'group' в порядке убывания значений
```

```python
# объявляем новую переменную, которая будет содержать дату; группу теста; суммарную выручку; количество уникальных посетителей
# в группе теста

visitorsAggregated = datesGroups.apply(
    lambda x: visitors[
        np.logical_and(
            visitors['date'] <= x['date'], visitors['group'] == x['group']
        )
    ].agg({'date': 'max', 'group': 'max', 'visitors': 'sum'}),
    axis=1,
).sort_values(by=['date', 'group']) # отсортируем столбцы 'date' и 'group' в порядке убывания значений
```

```python
# создадим новую переменную, где объеденим созданные датафреймы ('visitorsAggregated' и 'ordersAggregated') по столбцам 'date'
# и 'group'
cumulativeData = ordersAggregated.merge(visitorsAggregated, left_on=['date', 'group'], right_on = ['date', 'group'])
display(cumulativeData.head())
```

```python
# переименуем названия полей сформированного датасета
cumulativeData.columns = ['date',
                         'group',
                         'orders',
                         'buyers',
                         'revenue',
                         'visitors']
```

```python
# отберем показатели по каждой группе тестирования и запишем их в соответствующие переменные
cumulativeData_groupA = cumulativeData[cumulativeData['group'] == 'A']
cumulativeData_groupB = cumulativeData[cumulativeData['group'] == 'B']
```

```python
# построим графики накопления выручки по группам
plt.figure(figsize=(18, 6))
plt.plot(cumulativeData_groupA['date'], cumulativeData_groupA['revenue'], label = 'A')
plt.plot(cumulativeData_groupB['date'], cumulativeData_groupB['revenue'], label = 'B')
plt.grid()
plt.title('График накопления выручки по группам тестирования')
plt.xlabel('Даты')
plt.ylabel('Выручка, в млн. условных единиц')
plt.legend()
plt.show()
```

**Вывод**:

Можно увидеть рост выручки по двум группам тестирования: рост идет с переменным успехом и при наличии точки соприкосновения графиков двух групп - 13 августа 2019 года.

Затем заметно, как графики двух групп начинают расходиться друг от друга, а далее - 18 августа 2021 г. - происходит резкий рост выручки по группе "B": какой-то триггер повысил выручку по данной группе, в то время как группа "А" так же продолжала рост, но в "том же ритме" - без резких скачков.

### Построение графика кумулятивной среднего чека по группам

### Расчет статистической значимости различий между группами по "сырым" данным: конверсии и средний чек 

Прежде чем приступить непосредственно к расчетам статистической разности по группам нужно подготовить данные. Собрать в одну таблицу:
- дату;
- количество заказов на дату, отдельно по группам;
- выручку на дату, отдельно по группам;
- кумулятивное количество заказов на дату, отдельно по группам;
- кумулятивная выручка на дату, отдельно по группам;
- количество пользователей на дату, отдельно по группам;
- кумулятивное количество пользователей на дату, отдельно по группам;

```python
# количество посетителей по дням для группы А
visitorsADaily = visitors[visitors['group'] == 'A'][['date', 'visitors']]
visitorsADaily.columns = ['date', 'visitorsPerDateA']

visitorsACummulative = visitorsADaily.apply(
    lambda x: visitorsADaily[visitorsADaily['date'] <= x['date']].agg(
        {'date': 'max', 'visitorsPerDateA': 'sum'}
    ),
    axis=1,
)
visitorsACummulative.columns = ['date', 'visitorsCummulativeA']
```

```python
# количество посетителей по дням для группы B
visitorsBDaily = visitors[visitors['group'] == 'B'][['date', 'visitors']]
visitorsBDaily.columns = ['date', 'visitorsPerDateB']

visitorsBCummulative = visitorsBDaily.apply(
    lambda x: visitorsBDaily[visitorsBDaily['date'] <= x['date']].agg(
        {'date': 'max', 'visitorsPerDateB': 'sum'}
    ),
    axis=1,
)
visitorsBCummulative.columns = ['date', 'visitorsCummulativeB']
```

```python
# количество заказов по дням для группы А
ordersADaily = (
    orders[orders['group'] == 'A'][['date', 'transactionId', 'visitorId', 'revenue']]
    .groupby('date', as_index=False)
    .agg({'transactionId': pd.Series.nunique, 'revenue': 'sum'})
)
ordersADaily.columns = ['date', 'ordersPerDateA', 'revenuePerDateA']

ordersACummulative = ordersADaily.apply(
    lambda x: ordersADaily[ordersADaily['date'] <= x['date']].agg(
        {'date': 'max', 'ordersPerDateA': 'sum', 'revenuePerDateA': 'sum'}
    ),
    axis=1,
).sort_values(by=['date'])
ordersACummulative.columns = [
    'date',
    'ordersCummulativeA',
    'revenueCummulativeA',
]
```

```python
# количество заказов по дням для группы В
ordersBDaily = (
    orders[orders['group'] == 'B'][['date', 'transactionId', 'visitorId', 'revenue']]
    .groupby('date', as_index=False)
    .agg({'transactionId': pd.Series.nunique, 'revenue': 'sum'})
)
ordersBDaily.columns = ['date', 'ordersPerDateB', 'revenuePerDateB']

ordersBCummulative = ordersBDaily.apply(
    lambda x: ordersBDaily[ordersBDaily['date'] <= x['date']].agg(
        {'date': 'max', 'ordersPerDateB': 'sum', 'revenuePerDateB': 'sum'}
    ),
    axis=1,
).sort_values(by=['date'])
ordersBCummulative.columns = [
    'date',
    'ordersCummulativeB',
    'revenueCummulativeB',
]
```

```python
# соберем подготовленные данные в один датафрейм
data = (
    ordersADaily.merge(
        ordersBDaily, left_on='date', right_on='date', how='left'
    )
    .merge(ordersACummulative, left_on='date', right_on='date', how='left')
    .merge(ordersBCummulative, left_on='date', right_on='date', how='left')
    .merge(visitorsADaily, left_on='date', right_on='date', how='left')
    .merge(visitorsBDaily, left_on='date', right_on='date', how='left')
    .merge(visitorsACummulative, left_on='date', right_on='date', how='left')
    .merge(visitorsBCummulative, left_on='date', right_on='date', how='left')
)

print(data.head(5))
```

Создадим переменные ordersByUsersA и ordersByUsersB со столбцами ['userId', 'orders']. В них для пользователей, которые заказывали хотя бы 1 раз, укажем число совершённых заказов

```python
ordersByUsersA = (
    orders[orders['group'] == 'A']
    .groupby('visitorId', as_index=False)
    .agg({'transactionId': pd.Series.nunique})
)
ordersByUsersA.columns = ['userId', 'orders']

ordersByUsersB = (
    orders[orders['group'] == 'B']
    .groupby('visitorId', as_index=False)
    .agg({'transactionId': pd.Series.nunique})
)
ordersByUsersB.columns = ['userId', 'orders']
```

Объявим переменные sampleA и sampleB, в которых пользователям из разных групп будет соответствовать количество заказов. Тем, кто ничего не заказал, будут соответствовать нули. Это нужно, чтобы подготовить выборки к проверке критерием Манна-Уитни.

Формулируем гипотезы:

* H₀ - Вероятность того, что наблюдение из популяции B превышает наблюдение из популяции A равна 50%
* H₁ - Вероятность того, что наблюдение из популяции B превышает наблюдение из популяции A не равна 50%

```python
sampleA = pd.concat([ordersByUsersA['orders'],
                      pd.Series(0,index=np.arange(data['visitorsPerDateA'].sum() - len(ordersByUsersA['orders'])),
                                name='orders')],axis=0)

sampleB = pd.concat([ordersByUsersB['orders'],
                     pd.Series(0, index=np.arange(data['visitorsPerDateB'].sum() - len(ordersByUsersB['orders'])),
                               name='orders')],axis=0)
```

```python
# выведем на экран p-value и относительный прирост группы В к группе А
print("{0:.3f}".format(st.mannwhitneyu(sampleA, sampleB)[1]))

print("Относительный прирост группы В к группе А: {0:.3f}".format(sampleB.mean() / sampleA.mean() - 1))
```

**Вывод**:
- по «сырым» данным можно утверждать, что между группами существуют различия в конверсии.
- p-value получился меньше 0.05, это означает, что нулевую гипотезу о равенстве конверсий нужно отвергнуть. Значит конверсии не равны и более высокая конверсия группы В это не случайность.
- относительный прирост конверсии группы В по сравнению с группой А составил 13.8%

Теперь проверим статистическую значимость различий в среднем чеке между сегментами.

Формулируем гипотезы:

* H₀ - различий в среднем чеке между группами нет.
* H₁ - различия в среднем чеке между группами есть.

```python
print('{0:.3f}'.format(st.mannwhitneyu(orders[orders['group']=='A']['revenue'], orders[orders['group']=='B']['revenue'])[1]))
print('{0:.3f}'.format(orders[orders['group']=='B']['revenue'].mean()/orders[orders['group']=='A']['revenue'].mean()-1)) 
```

**Вывод**:
P-value значительно больше 0.05. Значит, причин отвергать нулевую гипотезу и считать, что в среднем чеке есть различия, нет. Средний чек группы B значительно выше среднего чека группы A.

### Расчет статистической значимости различий между группами по "очищенным" данным: конверсии и средний чек 

Примем за аномальных пользователей тех, кто совершил от 2 заказов или совершил заказ дороже 28 000 у.ед. Так мы уберём 5% пользователей с наибольшим числом заказов и от 1% до 5% пользователей с дорогими заказами. Сделаем срезы пользователей с числом заказов больше 2 — usersWithManyOrders и пользователей, совершивших заказы дороже 28 000 — usersWithExpensiveOrders. Объединим их в таблице abnormalUsers.

```python
# Узнаем, сколько всего аномальных пользователей атрибутом shape
usersWithManyOrders = pd.concat(
    [
        ordersByUsersA[ordersByUsersA['orders'] > 2]['userId'],
        ordersByUsersB[ordersByUsersB['orders'] > 2]['userId'],
    ],
    axis=0,
)
usersWithExpensiveOrders = orders[orders['revenue'] > 28000]['visitorId']
abnormalUsers = (
    pd.concat([usersWithManyOrders, usersWithExpensiveOrders], axis=0)
    .drop_duplicates()
    .sort_values()
)
print(abnormalUsers.head(5))
print(abnormalUsers.shape[0])
```

В общей сложности 74 аномальных пользователей.

Узнаем, как их действия повлияли на результаты теста.

Посчитаем статистическую значимость различий в среднем количестве заказов между группами теста по очищенным данным. Сначала подготовим выборки количества заказов по пользователям по группам теста и сформулируем гипотезы:

* H₀ - Вероятность того, что наблюдение из популяции B превышает наблюдение из популяции A равна 50%
* H₁ - Вероятность того, что наблюдение из популяции B превышает наблюдение из популяции A не равна 50%

```python
sampleAFiltered = pd.concat(
    [
        ordersByUsersA[
            np.logical_not(ordersByUsersA['userId'].isin(abnormalUsers))
        ]['orders'],
        pd.Series(
            0,
            index=np.arange(
                data['visitorsPerDateA'].sum() - len(ordersByUsersA['orders'])
            ),
            name='orders',
        ),
    ],
    axis=0,
)

sampleBFiltered = pd.concat(
    [
        ordersByUsersB[
            np.logical_not(ordersByUsersB['userId'].isin(abnormalUsers))
        ]['orders'],
        pd.Series(
            0,
            index=np.arange(
                data['visitorsPerDateB'].sum() - len(ordersByUsersB['orders'])
            ),
            name='orders',
        ),
    ],
    axis=0,
)
```

Применим статистический критерий Манна-Уитни к полученным выборкам:

```python
print('{0:.3f}'.format(st.mannwhitneyu(sampleAFiltered, sampleBFiltered)[1]))
print('Относительный прирост группы В к группе А: {0:.3f}'
      .format(sampleBFiltered.mean()/sampleAFiltered.mean()-1))
```

**Вывод**:
Как и в случае с сырыми данными, статистическая значимость достигнута. Сегмент `В` значительно лучше сегмента `А`. Отвергаем H₀.

Результаты по среднему количеству заказов практически не изменились. Произошло ли что-нибудь с результатами по среднему чеку?

Формулируем гипотезы для этого этапа:
* H₀ - Средний чек A равен среднему чеку группы B
* H₁ - Средний чек A не равен среднему чеку группы B

```python
print(
    '{0:.3f}'.format(
        st.mannwhitneyu(
            orders[
                np.logical_and(
                    orders['group'] == 'A',
                    np.logical_not(orders['visitorId'].isin(abnormalUsers)),
                )
            ]['revenue'],
            orders[
                np.logical_and(
                    orders['group'] == 'B',
                    np.logical_not(orders['visitorId'].isin(abnormalUsers)),
                )
            ]['revenue'],
        )[1]
    )
)

print(
    "{0:.3f}".format(
        orders[
            np.logical_and(
                orders['group'] == 'B',
                np.logical_not(orders['visitorId'].isin(abnormalUsers)),
            )
        ]['revenue'].mean()
        / orders[
            np.logical_and(
                orders['group'] == 'A',
                np.logical_not(orders['visitorId'].isin(abnormalUsers)),
            )
        ]['revenue'].mean()
        - 1
    )
)
```

**Вывод**:
- по «чистым» данным различий в среднем чеке нет.
- p-value получился сильно больше 0.05, это означает, что причин отвергать нулевую гипотезу о равенстве среднего чека нет.
- относительный прирост среднего чека в группе `В` составил -2%. Отвергаем нулевую гипотезу.
